### Truncating Data Set

**adult_new**: set represents individuals living in the United States, and some people's native countries only appear once or twice in the data set. This could create unstable income rates or uncessary outliers in the data set. To accomodate that, we are going to truncate the data set to only take into account with more than 25 entries.

```python
#check
len(adult_new)
```

```python
country_counts = adult_new['native_country'].value_counts()
filt_entry = country_counts[country_counts >= 25].index
print(len(filt_entry))
trunc_adult = adult_new[adult_new['native_country'].isin(filt_entry)].copy()
```

```python
#check
len(trunc_adult)
```

### Aggregating the Data Set
**adult_new (country) - PRIMARY **: We decided to aggregate the data set by the country because the adult data set contains individual date, but the country data set contains more general country data. We would not be able to meaningfully join the two tables without aggregating and take meaningiful statistics to compare.

**Specific Variable Decisions**

*income_rate:* Because of the conversion of income to a binary variable the mean represents the probability (or proportion) of individuals from that country earning more than $50K.

*education_num:* Education is measured in years and we aggregated by the mean to show the average education level of groups individuals in the adult set that are from other countries.

*hours_worked:* The hours worked influences the income, and averaging the hours per country helps to understand whether certain immigrant groups work more hours or if income reflects the difficulty of jobs.


```python
adult_country_aggregated = trunc_adult.groupby('native_country').agg(income_rate =('income','mean'),
 avg_ed =('education_num','mean'),
 avg_hrs=('hours_per_week','mean'),
 sample_size=('income','count')).reset_index()
```

```python
#check
print(adult_country_aggregated.shape)
```

### Joining the Datasets
Before we joined the two data sets, we had to make sure the country names in the two datasets would match each other because that is what we are joining the data sets on. To make them match we decided to strip both names in both datasets, remove any &/"and", and remove any spaces or dashes. Then we joined the adult data set with the country data set as well as the aggregated adult data set with the country data set. During the join, some of the countries in the adult data set do not exist in the countries data set, 7 of them to be specific. This is because they do not exist anymore or are considered parts of other countries. The unmerged entries will be excluded from any analysis involving the merged data set.

```python
#preparing for join
countries_new['country'] = (countries_new['country'].str.strip())
adult_country_aggregated['native_country'] = (adult_country_aggregated['native_country'].str.replace('-', ' ').str.replace('&', 'and').str.strip())
adult_new['native_country'] = (adult_country_aggregated['native_country'].str.replace('-', ' ').str.replace('&', 'and').str.strip())
country_fix = {
    'United-States': 'United States',
    'Puerto-Rico': 'United States',
    'El-Salvador': 'El Salvador',
    'Dominican-Republic': 'Dominican Republic',
    'Trinadad&Tobago': 'Trinidad & Tobago',
    'Holand-Netherlands': 'Netherlands',
    'Columbia': 'Colombia',
    'South': 'Korea, South',
    'Hong': 'Hong Kong',
    'Yugoslavia': 'Serbia',
    'England': 'United Kingdom',
    'Scotland': 'United Kingdom',
    'Outlying-US(Guam-USVI-etc)': 'United States',
    'United States of America' : 'United States'
}
adult_new['native_country'] = adult_new['native_country'].replace(country_fix)
adult_country_aggregated['native_country'] = adult_new['native_country'].replace(country_fix)

#join
merged_individuals = adult_new.merge(countries_new,
    left_on='native_country',
    right_on='country',
    how='left')
merged_country = adult_country_aggregated.merge(countries_new,
    left_on='native_country',
    right_on='country',
    how='inner')
```

### Further Aggregation
**adult_new (region)**: We decided to aggregate the data set by the region because it gives a similar understanding of the data set as by country, but it does give a slightly different view allowing for deeper understanding of regional differences between immigrant groups.

```python
region_agg = merged_country.groupby('region').agg(avg_income_rate=('income_rate','mean'),
    avg_literacy=('literacy','mean'),
    avg_population_density=('population_density','mean'),
    avg_agriculture=('agriculture','mean'),
    avg_industry=('industry','mean'),
    avg_service=('service','mean'),
    num_countries=('country','count')).reset_index()
```
